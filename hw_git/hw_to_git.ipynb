Напишите функции, которые будут выполнять следующие действия:
• Принимать три параметра (a: int, b: int, length: int) и возвращать список (list) длины length из равномерно
распределенной случайной величины между числами a и b
• Принимать созданный список (list) и считать количество уникальных значений переменной (строить частотный
ряд) в виде dict. Например, для списка [1, 2, 3, 1, 2, 2] будет возвращать {1 : 2, 2 : 3, 3 : 1}
• Считать минимум, среднюю, медиану, моду, максимум для значений списка

```python
import random
from collections import Counter
import statistics

def generate_random_list(a: int, b: int, length: int) -> list:
    """
    Возвращает список длины length из равномерно распределенной
    случайной величины между числами a и b.
    """
    #  random.uniform для непрерывной случайной величины.
    return [random.uniform(a, b) for _ in range(length)]

def build_frequency_dict(lst: list) -> dict:
    """
    Принимает список и строит частотный ряд в виде dict.
    """
    return dict(Counter(lst))

def calculate_statistics(lst: list) -> dict:
    """
    Считает минимум, среднюю, медиану, моду, максимум для значений списка.
    """
    if not lst:
        return {}

    try:
        mode_val = statistics.mode(lst)
    except statistics.StatisticsError:
        # Если уникальных мод несколько или список пуст
        mode_val = "Нет единственной моды"

    stats = {
        "минимум": min(lst),
        "среднее": statistics.mean(lst),
        "медиана": statistics.median(lst),
        "мода": mode_val,
        "максимум": max(lst)
    }
    return stats

# Пример использования:
test_list = [1, 2, 3, 1, 2, 2]
print(build_frequency_dict(test_list))
print(calculate_statistics(test_list))
```
