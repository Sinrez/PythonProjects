```python
import random
import numpy as np

def matrix(rows, cols):
  seed = random.randint(1, 2*(rows+cols))
  matrix_a = np.random.randint(1, seed, size=(rows, cols))
  matrix_b = np.random.randint(1, seed, size=(rows, cols))

  matrix_sum = matrix_a + matrix_b

  print("\nМатрица A:")
  print(matrix_a)

  print("\nМатрица B:")
  print(matrix_b)

  print("\nСумма матриц:")
  print(matrix_sum)

if __name__ == "__main__":
  try:
    m = int(input('Введите количество строк: '))
    n = int(input('Введите количество столбцов: '))
    matrix(m, n)
  except ValueError as ve:
      print(f'Должны быть введены только числа: {ve}!')
  except Exception as ex:
      print(f'Произошла ошибка: {ex}')
```
