```python
def decorator(func):
  """Декоратор, описывается до применения к нужной функции""""
    def wrapper():
        print("Подготовка к приветствию...")
        func()  # Здесь вызывается наша исходная функция
        print("Приветствие завершено.")
    return wrapper

@decorator #вызов декоратора
def say_hello():
    print("Привет!")

if __name__ == "__main__":
  say_hello()
```
