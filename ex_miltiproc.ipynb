```python
from multiprocessing import Process

def cpu_heavy_task():
    total = 0
    for i in range(1000000):
        total += i
    print(f"Задача завершена, итог: {total}")

if __name__ == "__main__":
    process1 = Process(target=cpu_heavy_task)
    process2 = Process(target=cpu_heavy_task)

    process1.start()
    process2.start()

    process1.join()
    process2.join()
```
