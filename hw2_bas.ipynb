Задача_2. Напишите функцию, которая принимает параметр – натуральное число N – и выводит на экран квадарт/прямоугольник из звездочек со стороной N.


```python
def sqrd(n: int) -> None:
    for _ in range(n):
        print('*' * n)

if __name__ == '__main__':
    sqrd(5)
```

Задача_3. Игральный кубик бросается три раза (выпадает три случайных значения). Сколько очков в среднем выпало?

```python
import random

def average_rolls(n=1000000):
    total = 0
    for _ in range(n):
        total += random.randint(1, 6) + random.randint(1, 6) + random.randint(1, 6)
    return total / n

print(average_rolls())  # ≈ 10.5
```

Задача_4. Напишите программу, которая вводит стороны треугольника и проверяет, можно ли его построить и вид построенного треугольника

```python

a = float(input("Введите сторону a: "))
b = float(input("Введите сторону b: "))
c = float(input("Введите сторону c: "))

# Проверка на существование
if a + b > c and a + c > b and b + c > a:
    print("Треугольник существует")

    # Определение вида
    if a == b == c:
        print("Треугольник равносторонний")
    elif a == b or a == c or b == c:
        print("Треугольник равнобедренный")
    else:
        print("Треугольник разносторонний")

    # Определение по углам
    sides = sorted([a, b, c])
    if sides[0]**2 + sides[1]**2 > sides[2]**2:
        print("Треугольник остроугольный")
    elif sides[0]**2 + sides[1]**2 == sides[2]**2:
        print("Треугольник прямоугольный")
    else:
        print("Треугольник тупоугольный")
else:
    print("Треугольник построить нельзя")
```

Задача_5. Определить равны ли введенные числа попарно между собой (вводим 3 целых числа), использовать логические переменные)

```python
a = int(input("Введите первое число: "))
b = int(input("Введите второе число: "))
c = int(input("Введите третье число: "))


equal_ab = (a == b)
equal_ac = (a == c)
equal_bc = (b == c)


if equal_ab and equal_ac and equal_bc:
    print("Все три числа равны между собой")
elif equal_ab:
    print("Первое и второе числа равны, но третье отличается")
elif equal_ac:
    print("Первое и третье числа равны, но второе отличается")
elif equal_bc:
    print("Второе и третье числа равны, но первое отличается")
else:
    print("Все три числа разные")
```

Задача_6. По номеру месяца определить количество дней в нем

```python
month = int(input("Введите номер месяца (1-12): "))
year = int(input("Введите год: "))

#проверки корректности
is_valid_month = 1 <= month <= 12
is_valid_year = year > 0

if not is_valid_month:
    print("Ошибка: номер месяца должен быть от 1 до 12")
elif not is_valid_year:
    print("Ошибка: год должен быть положительным числом")
else:
    # Проверка на високосный
    is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

    if month == 2:
        days = 29 if is_leap else 28
    elif month in (4, 6, 9, 11):
        days = 30
    else:
        days = 31

    months = ["", "январь", "февраль", "март", "апрель", "май", "июнь",
              "июль", "август", "сентябрь", "октябрь", "ноябрь", "декабрь"]

    print(f"{months[month].capitalize()} {year} года: {days} дней")
    if month == 2 and is_leap:
        print("(високосный год)")
```
