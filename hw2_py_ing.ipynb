*   Вывести имя и возраст самого младшего студента.
*   Найти имя студента с самой высокой средней оценкой.
*  Найти предмет, по которому самый высокий средний балл среди всех студентов.
* Удалить всех студентов, чей средний балл ниже 3.0.




```python
students = [
    {"name": "Иван", "age": 20, "grades": {"математика": 4, "физика": 5, "химия": 3}},
    {"name": "Мария", "age": 19, "grades": {"математика": 5, "физика": 5, "химия": 5}},
    {"name": "Пётр", "age": 21, "grades": {"математика": 3, "физика": 4, "химия": 2}},
    {"name": "Анна", "age": 22, "grades": {"математика": 5, "физика": 4, "химия": 4}},
]

def age_stud(students):
  min_age = students[0]["age"]
  min_name =''
  for i in range(len(students)):
    if min_age > students[i]["age"]:
      min_age = students[i]["age"]
      min_name = students[i]["name"]
  print(f'Самый младший студент: {min_name}, возраст: {min_age}')

def avg_score(grades):
  return sum(grades.values())/len(grades)

def max_scores_stud(students):
  max_avg_score = avg_score(students[0]["grades"])
  max_name =''
  for i in range(len(students)):
    if max_avg_score < avg_score(students[i]["grades"]):
      max_avg_score = avg_score(students[i]["grades"])
      max_name = students[i]["name"]
  print(f'Студент с самой высокой средней оценкой: {max_name}, средний балл: {max_avg_score:.2f}')

def max_scores_discipline(students):
    subjects = {}

    for student in students:
        for subject, grade in student["grades"].items():
            if subject not in subjects:
                subjects[subject] = []
            subjects[subject].append(grade)

    avg_scores = {}
    for subject, grades in subjects.items():
        avg_scores[subject] = sum(grades) / len(grades)

    max_subject, max_avg = max(avg_scores.items(), key=lambda x: x[1])
    print(f'Предмет с самым высоким средним баллом: {max_subject}, средний балл: {max_avg:.2f}')

def del_students(students):
  filtered_students = []
  for student in students:
      avg_grade = avg_score(student["grades"])
      if avg_grade > 3.0:
          filtered_students.append({
              "name": student["name"],
              "average": avg_grade
          })
  print("Студенты с баллом выше 3.0:")
  for student in filtered_students:
      print(f"{student['name']}: средний балл = {student['average']:.2f}")

if __name__ == '__main__':
  age_stud(students)
  max_scores_stud(students)
  max_scores_discipline(students)
  del_students(students)
```
