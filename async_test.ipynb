```python
from faker import Faker

fake = Faker("ru_RU")  # русская локаль


def get_random_name():
    # Отдаём новое случайное имя
    return fake.name()

# Генераторное выражение
name_gen = (get_random_name() for _ in range(5))

for name in name_gen:
    print(name)
```

```python
import asyncio

from faker import Faker

fake = Faker("ru_RU")  # русская локаль


async def get_random_name():
    await asyncio.sleep(0.5)
    return fake.name()


async def get_name_generator(count: int):
    for _ in range(count):
        yield await get_random_name()


async def main():
    async for name in get_name_generator(5):
        print(name)


#asyncio.run(main())

# В Colab - просто await вместо asyncio.run()
await main()
```

```python
import asyncio

from faker import Faker

fake = Faker("ru_RU")  # русская локаль


class AsyncRange:

    def __init__(self, stop: int, *, delay: int = 0):
        self.i = 0
        self.stop = stop
        self.delay = delay

    def __aiter__(self):
        # как __iter__, но для async
        return self

    async def __anext__(self):
        # как __next__, но асинхронный
        if self.i >= self.stop:
            raise StopAsyncIteration
        await asyncio.sleep(self.delay)  # имитируем асинхронную работу
        value = self.i
        self.i += 1
        return value


async def get_random_name():
    await asyncio.sleep(0.5)
    return fake.name()


async def main():
    name_gen = (await get_random_name() async for _ in AsyncRange(5))
    async for name in name_gen:
        print(name)


await main()
```
