В одном большом интернет-магазине книг менеджеры заметили, что количество проданных книг растёт, но им нужна аналитика для определения лучших продаж и отслеживания книг, которые продаются хуже других. Они обратились к тебе, чтобы ты написал программу, которая анализирует данные о продажах книг.

Тебе нужно:

Найти книги, которые продались более 100 экземпляров (filter).
Получить общую сумму продаж для каждой такой книги (map).
Вычислить общую сумму продаж всех отобранных книг (reduce).
Напишите программу, которая:

Принимает на вход список кортежей, где каждый кортеж содержит название книги и количество проданных экземпляров.
Использует filter для отфильтровки книг, которые продались более 100 экземпляров.
Использует map, чтобы рассчитать доход для каждой отфильтрованной книги (каждая книга стоит 500 рублей).
Использует reduce, чтобы найти общую сумму дохода от всех отфильтрованных книг.
Выводит названия книг с продажами более 100 экземпляров, доход по каждой книге и общую сумму дохода.
Цена каждой книги — 500 рублей.
Фильтруются книги, которые проданы в количестве более 100 экземпляров.
После фильтрации подсчитывается доход для каждой книги.
Итоговый доход выводится на экран.

Sample Input:

5
"Гарри Поттер" 150
"Война и мир" 50
"Мастер и Маргарита" 200
"Братья Карамазовы" 120
"Анна Каренина" 90

```python
from functools import reduce

n = int(input())
books = [tuple(input().lstrip('"').split('" ')) for _ in range(n)]

def is_valid_book(book):
    try:
        return int(book[1]) > 100
    except ValueError:
        return False

# Фильтруем книги и получаем только названия
result = [book[0] for book in filter(is_valid_book, books)]
print(f'Книги с продажами более 100 экземпляров: {result}')

def count_books(salary):
    return salary * 500

cost = [int(book[1]) for book in books if book[0] in result]
total_res = list(map(count_books, cost))
print(f'Доход по книгам: {total_res}')

def add_cost(x, y):
    return x + y

total_cost = reduce(add_cost, total_res)
print(f'Общий доход: {total_cost}')
```
