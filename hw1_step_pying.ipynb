```python
def str_to_words(strr : str):
  set_symb = {',', '.', '%', '"', "'", '!', '?', ';', ':'}
  for char in set_symb:
    strr = strr.replace(char, '')
  list_words = strr.split()
  print(f'Количество слов: {len(list_words)}')
  length_of_word = [len(word) for word in list_words]
  print(f'Средняя длина слова: {sum(length_of_word)/len(list_words):.1f}')


if __name__ == '__main__':
  inp_str = input('Введите строку: ').strip()
  str_to_words(inp_str)


```
