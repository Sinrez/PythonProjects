```python
import asyncio


async def main():
    queue = asyncio.Queue()  # Создаём очередь
    print(queue)
    await queue.put(1)  # Добавляем объект в очередь
    await queue.put(2)  # Добавляем объект в очередь
    await queue.put(3)  # Добавляем объект в очередь
    print(queue)


# asyncio.run(main())
# В Colab - просто await вместо asyncio.run()
await main()
```

когда мы вызвали метод put(3) — он «стал ждать» появления свободного места в очереди. Пока нет свободного места — задача не будет добавлена в очередь.
Программа не будет завершаться с ошибкой — мы будем просто ждать, пока появится свободное место в очереди.

```python
import asyncio


async def main():
    queue = asyncio.Queue(maxsize=2)  # Создаём очередь
    print(f'Элементов в очереди: {queue.qsize()}')
    await queue.put(1)  # Добавляем объект в очередь
    print(f'Элементов в очереди: {queue.qsize()}')
    await queue.put(2)  # Добавляем объект в очередь
    print(f'Элементов в очереди: {queue.qsize()}')
    await queue.get()  # Извлекаем объект из очереди
    print(f'Элементов в очереди: {queue.qsize()}')
    await queue.put(3)  # Добавляем объект в очередь
    print(f'Элементов в очереди: {queue.qsize()}')


await main()
```

```python
import asyncio
import random
from colorama import Fore, init

init(autoreset=True)  # чтобы цвета сбрасывались автоматически


async def producer(queue, task_count=1):
    """ Принимает какаую-то информацию и добавляет задачу в очередь """

    # Добавляем в очередь несколько задач
    for task_num in range(1, task_count + 1):
        await asyncio.sleep(random.uniform(0.1, 0.5))  # Имитация работы
        item = f"task-{task_num}"
        await queue.put(item)  # Положили в очередь
        print(f"{Fore.CYAN}Продюсер: добавил {item}")
    await queue.put(None)  # Флаг к завершению


async def consumer(queue):
    """ Берет задачу из очереди и выполняет её """
    while True:
        item = await queue.get()
        if item is None:  # Если получили None — прерываем цикл
            break
        print(f"{Fore.GREEN}Потребитель: взял {item}")
        await asyncio.sleep(random.uniform(0.2, 0.4))  # Обработка


async def main():
    queue = asyncio.Queue()
    # Создаём задачу для продюсера
    prod = asyncio.create_task(producer(queue, task_count=5))
    # Создаём задачу для консьюмера
    cons = asyncio.create_task(consumer(queue))
    # Ждём результат для сборки задач
    await asyncio.gather(prod, cons)


# asyncio.run(main())
# В Colab - просто await вместо asyncio.run()
await main()
```
